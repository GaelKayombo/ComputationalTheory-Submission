## Problem 1: Representing SHA-256 Data

This problem looks at how the different types of data used in SHA-256 can be represented in Python.Before working on the algorithm, I need to understand how it handles messages, 32bit words and 512 bit blocks.

### 1.1 Representing 32-bit words

SHA-256 uses 32 bit words when carrying out its calculations. Python integers can hold numbers of different sizes, so NumPy's uint32 is useful here because it represents unsigned 32 bit integers.

```python
import numpy as np

word = np.uint32(12345)

print(word)
print(word.dtype)
```

### 1.2 Representing sequences of 32-bit words

NumPy arrays allow us to store several 32bit words together. Using uint32 makes sure that each value is stored as an unsigned 32bit integer.

```python
words = np.array([1, 2, 3, 4], dtype=np.uint32)

print(words)
print(words.dtype)
print(len(words))
```
